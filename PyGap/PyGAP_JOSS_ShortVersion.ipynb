# Summary

`PyGap` is an open-source Python package providing tools and application programming interfaces (APIs) for data gap filling, signal separation, and structural analysis in spatio-temporal datasets. Incomplete observations and irregular sampling are pervasive issues in geophysics, climate science, and environmental monitoring. `PyGap` addresses these challenges by integrating two complementary methodologies: Least Squares Harmonic Estimation (LSHE) [@amiri2007least] and EigenSig [@li2019seismic]. Together, these methods enable robust missing data reconstruction, effective signal decomposition, and energy structure inspection within a modular framework.

# Statement of Need

Spatio-temporal geoscientific datasets frequently contain continuous missing observations due to instrument failures, transmission issues, or environmental constraints. Standard interpolation techniques (e.g., splines or kriging) often smooth out rapid physical variations or rely on strict stationarity assumptions. 

While LSHE accounts for white and colored noise structures, it assumes global stationarity, tends to suppress local non-harmonic features, and carries a high computational complexity $\mathcal{O}(n^3)$. Conversely, EigenSig relaxes global stationarity, preserves local features, and scales efficiently for large datasets. `PyGap` unifies LSHE and EigenSig into a single object-oriented library. This enables researchers to execute reproducible gap filling, perform statistical residual diagnostics, and process large-scale spatio-temporal series using a clean API.

# Functionality & Method Overview

`PyGap` operates directly on standard Python scientific data structures (e.g., `pandas.DataFrame`). Its core methodologies are streamlined as follows:

### Least Squares Harmonic Estimation (LSHE)

LSHE models an observed time series $y(t)$ as a linear combination of polynomial trends and $K$ detected angular frequencies $\omega_k$:

$$y = A \hat{x} + e = \hat{x}_0 + \hat{x}_1 t + \sum_{k=1}^K \left( \hat{a}_k \cos(\omega_k t) + \hat{b}_k \sin(\omega_k t) \right) + e$$

where $A$ is the design matrix containing trend and harmonic basis functions, $\hat{x}$ is the parameter vector estimated via generalized least squares (GLS), and $e$ is the noise vector. Frequencies are detected iteratively using Lomb–Scargle periodograms until Akaike Information Criterion (AIC) minimization is achieved. Missing epochs are interpolated by evaluating the globally fitted deterministic model $A \hat{x}$.

### EigenSig

EigenSig reshapes a time series of length $N$ into an $m \times p$ matrix $Y$ by segmenting it into uniform temporal epochs of length $p$. Singular Value Decomposition (SVD) decomposes this matrix into low-rank eigenspace components:

$$Y = U \Sigma V^T = \sum_{i=1}^r \sigma_i u_i v_i^T$$

where the right singular vectors $V$ form an orthonormal basis, and the left singular vectors $U$ represent trajectory coordinates. By interpolating these trajectory coordinates in the reduced eigenspace, missing temporal values are reconstructed upon inverse transformation without assuming global stationarity.

# Method Comparison

| Characteristic | EigenSig | LSHE | Hybrid (EigenSig+LSHE) |
| :--- | :--- | :--- | :--- |
| **Basis Functions** | Dynamic eigenvectors | Fixed Fourier series | LSHE within Eigenspace |
| **Feature Target** | Global and local variations | Global stationary patterns | Balanced global & local |
| **Computation** | Fast $\mathcal{O}(m \cdot p^2)$ | Heavy $\mathcal{O}(n^3)$ | Moderated |
| **Stationarity** | Not required | Required globally | Required locally in eigenspace |
